```python
from pathlib import Path
import re

REPO = Path(
    "/kaggle/working/Training-Large-Language-Models-to-Reason-and-Leverage-Search-Engines-with-Reinforcement-Learning"
)

demo_file = REPO / "scripts" / "demo_question.py"
code = demo_file.read_text()

# Add the output normalizer before main(), only once.
if "def normalize_model_output(" not in code:
    marker = "def main() -> None:"

    normalizer = r'''
def normalize_model_output(text: str) -> str:
    """Normalize common plain-text model output into the expected action tags."""

    # Remove common end-of-turn tokens from the visible response.
    text = text.replace("<|im_end|>", "").replace("<|endoftext|>", "").strip()

    # If the model already generated the required tags, preserve them.
    if "<search>" in text.lower() or "<answer>" in text.lower():
        return text

    # Convert the model's plain-text reasoning label.
    reasoning_match = re.search(
        r"Brief Reasoning:\s*(.*?)(?=Search Query:|Short Answer:|$)",
        text,
        re.IGNORECASE | re.DOTALL,
    )

    # Convert the model's plain-text search label.
    search_match = re.search(
        r"Search Query:\s*(.*?)(?=Short Answer:|$)",
        text,
        re.IGNORECASE | re.DOTALL,
    )

    # Convert the model's plain-text answer label.
    answer_match = re.search(
        r"Short Answer:\s*(.*?)(?:<\|im_end\|>|<\|endoftext\|>|$)",
        text,
        re.IGNORECASE | re.DOTALL,
    )

    parts = []

    if reasoning_match:
        reasoning = reasoning_match.group(1).strip()
        if reasoning:
            parts.append(f"<think>{reasoning}</think>")

    if search_match:
        query = search_match.group(1).strip()
        if query:
            parts.append(f"<search>{query}</search>")

    if answer_match:
        answer = answer_match.group(1).strip()
        if answer:
            parts.append(f"<answer>{answer}</answer>")

    # Do not silently discard an unexpected output.
    return "\n".join(parts) if parts else text
'''

    if marker not in code:
        raise RuntimeError("Could not find main() in demo_question.py")

    code = code.replace(marker, normalizer + "\n\n" + marker, 1)

# Preserve special tokens and normalize the generated response.
pattern = (
    r"return tokenizer\.decode\("
    r"new_tokens,\s*skip_special_tokens=(?:True|False)\s*\)"
)

replacement = (
    "return normalize_model_output("
    "tokenizer.decode(new_tokens, skip_special_tokens=False)"
    ")"
)

code, replacements = re.subn(pattern, replacement, code)

if replacements == 0:
    print("WARNING: The decode line was not changed automatically.")
    print("Inspect the generate() function in:", demo_file)
else:
    demo_file.write_text(code)
    print("Patched demo_question.py successfully.")
    print("Decode lines updated:", replacements)
```

```python
from pathlib import Path

rollout_file = REPO / "src" / "search_rl" / "rollout.py"
code = rollout_file.read_text()

start = code.find("    def run(self, question: str, generate: Callable[[str], str]) -> Trajectory:")

if start == -1:
    raise RuntimeError("Could not find SearchEnvironment.run() in rollout.py")

# The run() method is the final method in the current rollout.py.
new_method = '''    def run(self, question: str, generate: Callable[[str], str]) -> Trajectory:
        started = time.perf_counter()
        prompt = self.build_prompt(question)
        full_text = ""
        events: list[SearchEvent] = []
        answer = ""
        termination = "no_valid_action"

        for _ in range(self.max_searches + 1):
            generated = generate(prompt)

            # Look for the first actionable tag in this generation.
            search_match = SEARCH_PATTERN.search(generated)
            answer_match = ANSWER_PATTERN.search(generated)

            # If a search occurs before the answer, execute the search first.
            # Ignore generated text after the search action for this turn.
            if (
                search_match is not None
                and len(events) < self.max_searches
                and (
                    answer_match is None
                    or search_match.start() < answer_match.start()
                )
            ):
                full_text += generated[:search_match.end()]
                query = search_match.group(1).strip()

                if not query:
                    termination = "no_valid_action"
                    answer = full_text.strip()
                    break

                passages = self.retriever.search(query, top_k=self.top_k)
                events.append(SearchEvent(query=query, passages=passages))

                # Ask the model to continue using the retrieved evidence.
                evidence = self._evidence_block(passages)
                prompt = self.build_prompt(question, evidence)
                continue

            # Accept an answer only after any earlier search action was handled.
            if answer_match is not None:
                full_text += generated
                answer = answer_match.group(1).strip()
                termination = "answer"
                break

            # No valid action was generated.
            full_text += generated
            termination = "no_valid_action"
            answer = full_text.strip()
            break

        else:
            termination = "search_limit"
            answer = full_text.strip()

        return Trajectory(
            question=question,
            text=full_text,
            answer=answer,
            searches=events,
            termination=termination,
            latency_seconds=time.perf_counter() - started,
        )
'''

code = code[:start] + new_method + "\n"
rollout_file.write_text(code)

print("Patched rollout.py successfully.")
print("Search actions now take priority when they appear before an answer.")
```

```python
from pathlib import Path

REPO = Path(
    "/kaggle/working/Training-Large-Language-Models-to-Reason-and-Leverage-Search-Engines-with-Reinforcement-Learning"
)

demo_file = REPO / "scripts" / "demo_question.py"
code = demo_file.read_text()

# Remove the misplaced import.
code = code.replace(
    '"""Show one end-user Search-RL trajectory, including model and tool steps.\n',
    '"""Show one end-user Search-RL trajectory, including model and tool steps.\n',
    1
)

# Remove standalone import re lines, then add it after the future import.
lines = [
    line for line in code.splitlines()
    if line.strip() != "import re"
]

code = "\n".join(lines) + "\n"

future_import = "from __future__ import annotations"

if future_import not in code:
    raise RuntimeError("Could not find the future import.")

if "import re" not in code:
    code = code.replace(
        future_import,
        future_import + "\n\nimport re",
        1
    )

demo_file.write_text(code)

print("Fixed import order.")
print("\n".join(demo_file.read_text().splitlines()[:18]))
```
