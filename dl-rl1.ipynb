```python
from pathlib import Path
import shutil

dataset_root = Path(
    "/kaggle/input/datasets/shubhamdineshjoshi/checkpoints-and-data"
)

project_root = Path(
    "/kaggle/working/Training-Large-Language-Models-to-Reason-and-Leverage-Search-Engines-with-Reinforcement-Learning"
)

print("Dataset exists:", dataset_root.exists())
print("Project exists:", project_root.exists())

(project_root / "data" / "processed").mkdir(parents=True, exist_ok=True)
(project_root / "artifacts").mkdir(parents=True, exist_ok=True)

for filename in ["train.jsonl", "validation.jsonl", "passages.jsonl"]:
    source = dataset_root / "data" / "processed" / filename
    destination = project_root / "data" / "processed" / filename

    if source.exists():
        shutil.copy2(source, destination)
        print("Copied:", destination)
    else:
        print("Missing:", source)

source_artifacts = dataset_root / "artifacts"
destination_artifacts = project_root / "artifacts"

if source_artifacts.exists():
    shutil.copytree(
        source_artifacts,
        destination_artifacts,
        dirs_exist_ok=True
    )
    print("Artifacts restored")
else:
    print("Artifacts folder missing")
```

```python
from pathlib import Path
import shutil
import json

project = Path("/kaggle/working")
save_dir = project / "search_rl_checkpoint_before_5000"

if save_dir.exists():
    shutil.rmtree(save_dir)

save_dir.mkdir(parents=True)

# Save processed data
data_source = project / "data" / "processed"
data_target = save_dir / "data" / "processed"
data_target.mkdir(parents=True)

for filename in ["train.jsonl", "validation.jsonl", "passages.jsonl"]:
    source = data_source / filename
    if source.exists():
        shutil.copy2(source, data_target / filename)

# Save the latest format adapter
for adapter_name in [
    "qwen05b-format-warmup-v4",
    "qwen05b-format-warmup-v3",
    "qwen05b-search-rl-500-50",
]:
    source = project / "artifacts" / adapter_name
    if source.exists():
        shutil.copytree(
            source,
            save_dir / "artifacts" / adapter_name,
            dirs_exist_ok=True,
        )

manifest = {
    "stage": "before 5000-example training",
    "latest_format_adapter": "qwen05b-format-warmup-v4",
    "status": "format protocol not yet reliable",
    "do_not_start_5000_training": True,
    "known_issues": [
        "Model does not consistently emit think/search/answer tags",
        "BM25 passages are weak and often irrelevant",
        "Some FP16 warm-up steps show nan grad_norm",
    ],
}

(save_dir / "manifest.json").write_text(
    json.dumps(manifest, indent=2),
    encoding="utf-8",
)

archive = shutil.make_archive(
    "/kaggle/working/search_rl_before_5000",
    "zip",
    root_dir="/kaggle/working",
    base_dir="search_rl_checkpoint_before_5000",
)

print("Created:", archive)
```
