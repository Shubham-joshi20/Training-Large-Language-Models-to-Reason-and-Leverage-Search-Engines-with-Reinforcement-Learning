```python
import pandas as pd

base_path = "/kaggle/input/datasets/frankossai/natural-questions-dataset/Natural-Questions-Base.csv"
filtered_path = "/kaggle/input/datasets/frankossai/natural-questions-dataset/Natural-Questions-Filtered.csv"

for path in [filtered_path, base_path]:
    df = pd.read_csv(path)
    print("\nFILE:", path)
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    display(df.head(1).T)
```

```python
df = pd.read_csv(base_path)

possible_columns = [
    "document_text",
    "context",
    "passage",
    "long_answer",
    "document",
    "text"
]

for column in possible_columns:
    if column in df.columns:
        print(column, "non-empty values:", df[column].notna().sum())
        print(df[column].dropna().iloc[0][:500])
```
